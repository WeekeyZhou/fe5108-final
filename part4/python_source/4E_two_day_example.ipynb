# Part 4E: Two-day example (+10% then -9%)

**Question.** NVDA is up 10% one day and down 9% the next. What are the two-day returns of NVDA and NVDL (ignoring daily drag)? How does this relate to the factsheet warning that the Fund will lose money if the underlying stock is flat, and may lose money even when the stock rises over a period longer than a single day?

**Setup.** NVDL earns exactly twice the NVDA return each day, then compounds.

```python
import pandas as pd
import numpy as np
pd.options.display.float_format = "{:,.4%}".format

def two_day(r1, r2, lev=2):
    nvda = (1 + r1) * (1 + r2) - 1
    etf = (1 + lev * r1) * (1 + lev * r2) - 1
    return nvda, etf

r1, r2 = 0.10, -0.09
nvda2, nvdl2 = two_day(r1, r2)

path = pd.DataFrame({
    "NVDA daily": [r1, r2],
    "NVDA value ($1 start)": [1 + r1, (1 + r1) * (1 + r2)],
    "NVDL daily (2x)": [2 * r1, 2 * r2],
    "NVDL value ($1 start)": [1 + 2 * r1, (1 + 2 * r1) * (1 + 2 * r2)],
}, index=["Day 1", "Day 2"])
display(path)
print(f"Two-day return, NVDA: {nvda2:.2%}")
print(f"Two-day return, NVDL: {nvdl2:.2%}   (naive 2 x NVDA two-day return would be {2 * nvda2:.2%})")
```

```python
# A flat stock: +10% on day 1, then the exact fall that returns NVDA to its starting price
r2_flat = 1 / 1.10 - 1
nvda_f, nvdl_f = two_day(0.10, r2_flat)
print(f"Flat case: day-2 NVDA return {r2_flat:.4%}, NVDA two-day {nvda_f:.4%}, NVDL two-day {nvdl_f:.4%}")

# Closed form for the gap: NVDL - 2*NVDA over two days = 2*r1*r2
print(f"Gap in the +10%/-9% example: 2*r1*r2 = {2 * r1 * r2:.2%}  (NVDL - 2xNVDA = {nvdl2 - 2 * nvda2:.2%})")
```
